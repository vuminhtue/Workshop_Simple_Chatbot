```python
import ollama

def chatbot(question,temperature=0.7, model='gemma3:1b'):
   # Send the question to Ollama with gemma3:1b model
   response = ollama.chat(
       model=model,
       messages=[{'role': 'user', 'content': question}],
       options=ollama.Options(temperature=temperature)
   )
   # Return the answer from the model
   return response['message']['content']
```

```python
chatbot("What is the capital of France?")
```
